```python
import random
from gtts import gTTS
import os

# Function to generate a random story based on user input
def generate_story(character_name, setting, genre):
    # Predefined story elements
    introductions = [
        f"Once upon a time in {setting}, there lived a brave soul named {character_name}.",
        f"In the {setting}, {character_name} was known for their courage and kindness.",
    ]
    conflicts = [
        f"One day, a great challenge appeared. A mysterious force threatened {setting}, and only {character_name} could stop it.",
        f"Suddenly, a dark cloud filled the sky, and {character_name} had to journey to save the land from the looming threat."
    ]
    resolutions = [
        f"After many adventures, {character_name} triumphed over the danger and brought peace back to {setting}.",
        f"With courage and wisdom, {character_name} outsmarted the evil forces and restored balance to {setting}."
    ]

    # Randomly select parts of the story
    intro = random.choice(introductions)
    conflict = random.choice(conflicts)
    resolution = random.choice(resolutions)

    # Return the complete story
    story = f"{intro} {conflict} {resolution} It was a tale that became a legend in {setting}."

    return story

# Function to convert text to speech and play it
def text_to_speech(text):
    tts = gTTS(text)
    tts.save("story.mp3")
    os.system("start story.mp3")  # This will play the MP3 on most systems (Windows)

# Function to run the storytelling bot
def storytelling_bot():
    print("Welcome to the Storytelling Bot!")

    # Get user input for the story elements
    character_name = input("What is the name of the main character? ")
    setting = input("Where is the story set (e.g., a magical forest, a faraway kingdom)? ")
    genre = input("What genre is the story (e.g., fantasy, adventure, mystery)? ")

    # Generate the story
    story = generate_story(character_name, setting, genre)

    # Print the generated story
    print("\nHere is your story:\n")
    print(story)

    # Narrate the story using TTS
    print("\nNow, let me narrate it for you...")
    text_to_speech(story)

# Run the bot
storytelling_bot()
```

```python
import os
from gtts import gTTS
from moviepy.editor import *
from PIL import Image, ImageDraw, ImageFont

def text_to_audio(text, audio_filename="output.mp3"):

    tts = gTTS(text)
    tts.save(audio_filename)

def create_image_with_text(text, image_filename="background.png"):


    width, height = 1920, 1080
    image = Image.new('RGB', (width, height), color=(255, 255, 255))


    draw = ImageDraw.Draw(image)
    font = ImageFont.load_default()


    # Use textbbox instead of textsize if using Pillow < 9.3.0
    # text_size = draw.textsize(text, font=font) #Commented out the line causing the error.
    text_size = draw.textbbox((0,0),text, font=font) #Use textbbox instead of textsize if using Pillow < 9.3.0. textbbox returns (x0, y0, x1, y1)
    text_width = text_size[2]-text_size[0] #Calculate width of the text
    text_height = text_size[3]-text_size[1] #Calculate height of the text


    text_position = ((width - text_width) // 2, (height - text_height) // 2) # Use text width and height to position the text.


    draw.text(text_position, text, fill="black", font=font)


    image.save(image_filename)

def create_video_from_audio_and_text(audio_filename, video_filename="output_video.mp4", text=""):

    audio_clip = AudioFileClip(audio_filename)
    duration = audio_clip.duration


    background_image = "background.png"
    create_image_with_text(text, background_image)


    image_clip = ImageClip(background_image).set_duration(duration)
    image_clip = image_clip.set_audio(audio_clip)


    image_clip.write_videofile(video_filename, fps=24)

def text_to_video(text, video_filename="output_video.mp4"):
    """Converts text to speech and generates a video."""
    audio_filename = "output.mp3"


    text_to_audio(text, audio_filename)


    create_video_from_audio_and_text(audio_filename, video_filename, text)


    os.remove(audio_filename)
    os.remove("background.png")

text = input("Enter the text you want to convert to video: ")
text_to_video(text)
```

```python
import os
from gtts import gTTS
from moviepy.editor import *
from PIL import Image, ImageDraw, ImageFont
from IPython.display import Video

def text_to_audio(text, audio_filename="output.mp3"):
    tts = gTTS(text)
    tts.save(audio_filename)

def create_image_with_text(text, image_filename="background.png", width=1920, height=1080, font_size=48):
    image = Image.new('RGB', (width, height), color=(255, 255, 255))
    draw = ImageDraw.Draw(image)

    try:
        font = ImageFont.truetype("arial.ttf", font_size)
    except OSError:
        font = ImageFont.load_default()
        print("Warning: arial.ttf not found. Using default font.")
    text_size = draw.textbbox((0, 0), text, font=font)
    text_width = text_size[2] - text_size[0]
    text_height = text_size[3] - text_size[1]
    text_position = ((width - text_width) // 2, (height - text_height) // 2)
    draw.text(text_position, text, fill="black", font=font)
    image.save(image_filename)

def create_video_from_audio_and_text(audio_filename, video_filename="output_video.mp4", text=""):
    audio_clip = AudioFileClip(audio_filename)
    duration = audio_clip.duration
    background_image = "background.png"

    create_image_with_text(text, background_image)

    image_clip = ImageClip(background_image).set_duration(duration)


    print("Audio duration:", duration)


    video_clip = image_clip.set_audio(audio_clip)


    video_clip.write_videofile(video_filename, fps=24)

def text_to_video(text, video_filename="output_video.mp4"):
    audio_filename = "output.mp3"
    text_to_audio(text, audio_filename)
    create_video_from_audio_and_text(audio_filename, video_filename, text)
    os.remove(audio_filename)
    os.remove("background.png")
    return Video(video_filename, embed=True, html_attributes="controls muted autoplay")

!wget -O arial.ttf "https://github.com/google/fonts/raw/main/ofl/ariblk/Arimo-Bold.ttf"

!apt-get update && apt-get install -y ffmpeg

text = input("Enter the text you want to convert to video: ")
video_output = text_to_video(text)
display(video_output)
```

```python
from diffusers import DiffusionPipeline
import torch
import numpy as np
import imageio
from PIL import Image

pipe = DiffusionPipeline.from_pretrained(
    "damo-vilab/text-to-video-ms-1.7b",
    torch_dtype=torch.float16,
    variant="fp16"
).to("cuda")

prompt = "modi dancing with a girl"
video_batches = pipe(prompt, num_inference_steps=50).frames

target_resolution = (512, 512)

processed_frames = []

for i, batch in enumerate(video_batches):
    for j, frame in enumerate(batch):

        if frame.dtype != np.uint8:
            frame = (frame * 255).clip(0, 255).astype(np.uint8)


        if frame.ndim == 2:
            frame = np.stack([frame] * 3, axis=-1)
        elif frame.ndim == 3 and frame.shape[2] == 1:
            frame = np.repeat(frame, 3, axis=2)
        elif frame.ndim == 3 and frame.shape[2] > 4:
            frame = frame[:, :, :3]

        image = Image.fromarray(frame)
        image = image.resize(target_resolution, Image.LANCZOS)
        processed_frames.append(np.array(image))

print(f"✅ Total processed frames: {len(processed_frames)}")


output_path = "high_quality_video.mp4"
writer = imageio.get_writer(
    output_path,
    fps=8,
    codec='libx264',
    bitrate="5M",
    quality=10
)

for frame in processed_frames:
    writer.append_data(frame)
writer.close()

print(f"🎥 High-quality video saved as {output_path}")
```
