```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

img = cv2.imread("Naruto .jpg")

img_rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

h, w, c = img_rgb.shape

t_x, t_y = 500, 200

translated_img = np.zeros_like(img_rgb)

for i in range(h):
    for j in range(w):
        
        new_i = i + t_y
        new_j = j + t_x

        if 0 <= new_i < h and 0 <= new_j < w:
            translated_img[new_i, new_j] = img_rgb[i, j]
            
plt.title("TRANSLATION IMAGE")
plt.imshow(translated_img)
plt.axis("off")
plt.show()
```

```python
import math
ang_rad = math.radians(45)

ctr_x, ctr_y = w // 2, h // 2

rot_img = np.zeros_like(img_rgb)

for i in range(h):
    for j in range(w):
        
        x_shift = j - ctr_x
        y_shift = i - ctr_y
        
        
        new_x = int(x_shift * math.cos(ang_rad) - y_shift * math.sin(ang_rad) + ctr_x)
        new_y = int(x_shift * math.sin(ang_rad) + y_shift * math.cos(ang_rad) + ctr_y)
        
        if 0 <= new_x < w and 0 <= new_y < h:
            rot_img[new_y, new_x] = img_rgb[i, j]
plt.title("Rotation image")
plt.imshow(rot_img)
plt.axis("off")
plt.show()
    
```

```python
new_h = 400  
new_w = 600  
scale_y = new_h / h
scale_x = new_w / w

scaled_img = np.zeros((new_h, new_w, 3), dtype=np.uint8)

for i in range(new_h):
    for j in range(new_w):
        
        org_i = int(i / scale_y)
        org_j = int(j / scale_x)

        org_i = min(org_i, h- 1)
        org_j = min(org_j, w - 1)

        scaled_img[i, j] = img_rgb[org_i, org_j]
plt.figure(figsize=(12,8))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(scaled_img)
plt.title('Scaled Image')

plt.show()
scaled_img.shape
```

```python

Sh_x = 0.5 
Sh_y = 0.3
sheared_img = np.zeros_like(img_rgb)

for i in range(h):
    for j in range(w):
       
        new_x = int(j + Sh_x * i)  
        new_y = i  

        if 0 <= new_x < w and 0 <= new_y < h:
            sheared_img[new_y, new_x] = img_rgb[i, j]

plt.figure(figsize=(12,8))
plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis()

plt.subplot(1, 2, 2)
plt.imshow(sheared_img)
plt.title("Sheared Image (Horizontal)")
plt.axis()

plt.show()
```

```python
sheared_img_vertical = np.zeros_like(img_rgb)

for y in range(h):
    for x in range(w):
        new_y = int(y + Sh_y * x)
        
        if 0 <= new_y < h:
            sheared_img_vertical[new_y, x] = img_rgb[y, x]

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis()

plt.subplot(1, 2, 2)
plt.imshow(sheared_img_vertical)
plt.title("Sheared Image (Vertical)")


```
