```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
image = Image.open('colorful.png')  
image = image.convert('RGB') 
rgb_array = np.array(image) / 255.0
cmy_array = 1 - rgb_array
print(cmy_array[:,:,2])
cmy_array = (cmy_array * 255).astype(np.uint8)
cmy_image = Image.fromarray(cmy_array)
plt.imshow(cmy_array)
plt.axis('off') 
plt.show()
cmy_image.save('colorful.png')
```

```python
c_array=cmy_array
c_array = (c_array * 255).astype(np.uint8)
c_array[:,:,0]=1
m_array=cmy_array
m_array = (m_array * 255).astype(np.uint8)
m_array[:,:,1]=1
y_array=cmy_array
y_array = (y_array * 255).astype(np.uint8)
y_array[:,:,2]=1
cm_array=cmy_array
cm_array = (cm_array * 255).astype(np.uint8)
cm_array[:,:,0:2]=1
my_array=cmy_array
my_array = (my_array * 255).astype(np.uint8)
my_array[:,:,1:3]=1
cy_array=cmy_array
cy_array = (cy_array * 255).astype(np.uint8)
cy_array[:,:,0:3:2]=1
a=np.hstack((cmy_array,c_array,m_array,y_array,my_array,cy_array,cm_array))
plt.imshow(a)
plt.axis('off') 
plt.show()
```
